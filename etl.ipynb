## Extract

```python
import pandas as pd

from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("Gaming_Academic_Performance.csv")
```

## Transform

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
le = LabelEncoder()

df["gender"] = le.fit_transform(df["gender"])

stress_map = {
    "Low":0,
    "Medium":1,
    "High":2
}

df["stress_level"] = df["stress_level"].map(stress_map)
```

```python
df
```

## Load

```python
df.to_csv("Gaming_Academic_Performance_ETL.csv", index=False)
```

## Data untuk Data Warehouse & Machine Learning

```python
df_dw = df.copy()
df_ml = df.copy()
```

## Data Warehouse

```python
df_dw["gender_id"] = df_dw["gender"]
```

```python
genre_map = {
    "Casual":1,
    "FPS":2,
    "RPG":3
}

df_dw["genre_id"] = df_dw["gaming_genre"].map(genre_map)
```

```python
df_dw["stress_id"] = df_dw["stress_level"]
```

```python
df_dw = df_dw.drop(columns=["gender", "gaming_genre", "stress_level"])
```

```python
df_dw = df_dw[
    [
        "student_id",
        "gender_id",
        "genre_id",
        "stress_id",
        "age",
        "gaming_hours",
        "study_hours",
        "sleep_hours",
        "attendance",
        "social_activity",
        "device_usage",
        "reaction_time_ms",
        "addiction_score",
        "grades"
    ]
]
```

```python
df_dw
```

```python
df_dw.to_csv("Gaming_Academic_Performance_DW.csv", index=False)
```

## Data Machine Learning

```python
df_ml = pd.get_dummies(
    df_ml,
    columns=["gaming_genre"],
    drop_first=False
)
```

```python
df_ml.drop(columns=["student_id"], inplace=True)
```

```python
df_ml
```

```python
df_ml.to_csv("Gaming_Academic_Performance_DM.csv", index=False)
```
